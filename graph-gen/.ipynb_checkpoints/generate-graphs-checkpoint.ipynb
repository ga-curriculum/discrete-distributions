```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

```python
GA_RED = '#E51B24'
```

```python
X = stats.randint(1, 7)
xs = np.arange(-1, 9)
```

```python
plt.figure(figsize=(6, 3))
plt.bar(xs, X.pmf(xs), color=GA_RED)
plt.xticks(xs)
plt.yticks([1/6], labels=['1/6'])
plt.ylim(0, 0.22)
plt.ylabel('Probability')
plt.xlabel('Die roll')
plt.tight_layout()
plt.savefig('graphs/uniform-intro.png', dpi=192);
```
