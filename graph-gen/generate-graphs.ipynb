```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

```python
GA_RED = '#E51B24'
```

```python
X = stats.randint(1, 7)
xs = np.arange(-1, 9)
```

```python
plt.figure(figsize=(6, 3))
plt.bar(xs, X.pmf(xs), color=GA_RED)
plt.xticks(xs)
plt.yticks([1/6], labels=['1/6'])
plt.ylim(0, 0.22)
plt.ylabel('Probability')
plt.xlabel('Die roll')
plt.tight_layout()
plt.savefig('graphs/uniform-intro.png', dpi=192);
```

```python
X = stats.randint(1, 7)
xs = np.arange(0, 8)
plt.figure(figsize=(6, 3))
plt.bar(xs, X.pmf(xs), color=GA_RED)
plt.xticks(xs, labels=['', 'a', '', '', '', '', 'b', ''])
plt.yticks([1/6], labels=['$\\frac{1}{b - a}$'])
plt.ylim(0, 0.22)
plt.tight_layout()
plt.savefig('graphs/uniform-pdf.png', dpi=192);
```

```python
X = stats.randint(1, 7)
xs = np.linspace(-1, 8, 1000)
plt.figure(figsize=(6, 3))
plt.plot(xs, X.cdf(xs), color=GA_RED)
plt.xticks(np.arange(-1, 9))
plt.yticks(np.arange(0, 7)/6, labels=['0'] + [f'{i}/6' for i in range(1, 6)] + ['1'])
plt.tight_layout()
plt.savefig('graphs/uniform-cdf.png', dpi=192);
```

```python
n = 8
p = 0.3
X = stats.binom(n, p)
xs = np.arange(0, n + 1)

plt.bar(xs, X.pmf(xs), color=GA_RED)
plt.xticks(xs)
plt.tight_layout()
plt.savefig('graphs/binom-pdf.png', dpi=192);
```

```python
X = stats.poisson(mu=3)
xs = np.arange(0, 13)

plt.figure(figsize=(4, 4))
plt.bar(xs, X.pmf(xs), color=GA_RED)
plt.xticks([])
plt.yticks([])
plt.tight_layout()
plt.savefig('graphs/poisson-pdf.png', dpi=192);
```
